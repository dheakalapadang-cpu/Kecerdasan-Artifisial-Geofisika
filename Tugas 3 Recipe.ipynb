RECIPE III — Model Difusi Panas Satu Dimensi

1. Problem

Mensimulasikan perubahan distribusi suhu terhadap ruang dan waktu menggunakan model difusi panas satu dimensi.

Input:

- "INITIAL_PROFILE": "[INITIAL_PROFILE]"
- "ALPHA": "[ALPHA]"
- "BOUNDARIES": "[BOUNDARIES]"
- "DZ": "[DZ]"
- "DT": "[DT]"

Output:

- Profil suhu: "[OUTPUT_PROFILE]"
- Grafik: "[FIGURE_PATH]"
- Run record: "[RUN_PATH]"

2. Minimal Theory

Persamaan difusi panas satu dimensi dinyatakan sebagai:

[
\frac{\partial T}{\partial t}

\alpha \frac{\partial^2 T}{\partial z^2}
]

Keterangan:

- (T) = suhu (K atau °C untuk perbedaan suhu).
- (t) = waktu (s).
- (z) = posisi atau kedalaman (m).
- (\alpha) = difusivitas termal ((\mathrm{m^2/s})).

Untuk skema eksplisit beda hingga dengan grid ruang seragam, parameter kestabilan didefinisikan sebagai:

[
r = \frac{\alpha \Delta t}{(\Delta z)^2}
]

Untuk persamaan difusi satu dimensi dengan skema eksplisit standar, syarat kestabilan numerik adalah:

[
r \leq \frac{1}{2}
]

Jika nilai (r) lebih besar dari (1/2), program harus memberikan peringatan atau menghentikan simulasi sesuai kebijakan validasi yang ditetapkan.

Hasil simulasi juga bergantung pada profil suhu awal, kondisi batas, ukuran grid, dan durasi simulasi.

Berikan peringatan jika parameter tidak valid. Jangan mengklaim kalibrasi, konvergensi, atau keberhasilan eksekusi tanpa bukti pengujian.

# Analisis Curah Hujan dan Deteksi Hujan Ekstrem

## 1. Problem

Melakukan analisis QA/QC dan analisis statistik pada data curah hujan untuk mengidentifikasi pola curah hujan, nilai ekstrem, data kosong, dan perubahan curah hujan berdasarkan waktu.

Program menggunakan data curah hujan harian dari stasiun pengamatan. Data dianalisis untuk memperoleh statistik deskriptif dan mengidentifikasi kejadian hujan ekstrem berdasarkan ambang batas yang ditentukan pengguna.

**Input:**
- `RAINFALL_PATH`: `[RAINFALL_PATH]`
- `DATE_COLUMN`: `[DATE_COLUMN]`
- `RAINFALL_COLUMN`: `[RAINFALL_COLUMN]`
- `EXTREME_THRESHOLD`: `[EXTREME_THRESHOLD]`

**Output:**
- Data hasil analisis CSV: `[OUTPUT_CSV_PATH]`
- Laporan Markdown: `[REPORT_PATH]`
- Grafik curah hujan: `[FIGURE_PATH]`
- Run record JSON: `[RUN_PATH]`

## 2. Minimal Theory

Curah hujan merupakan jumlah presipitasi yang jatuh pada suatu wilayah dalam periode tertentu. Data curah hujan dapat digunakan untuk mempelajari variasi hujan, mengidentifikasi kejadian hujan ekstrem, dan mendukung analisis risiko bencana hidrometeorologi.

Analisis dilakukan menggunakan statistik deskriptif, pemeriksaan kualitas data, dan identifikasi nilai yang melampaui ambang batas.

**Rata-rata curah hujan:**

\[
\bar{R}=\frac{1}{n}\sum_{i=1}^{n}R_i
\]

**Curah hujan maksimum:**

\[
R_{\max}=\max(R_1,R_2,\ldots,R_n)
\]

**Jumlah kejadian hujan yang melampaui ambang batas:**

\[
N_e=\sum_{i=1}^{n}I(R_i\geq R_{\text{threshold}})
\]

Keterangan:
- \(R_i\) = curah hujan pada hari ke-\(i\), dalam mm/hari.
- \(\bar{R}\) = rata-rata curah hujan.
- \(R_{\max}\) = curah hujan maksimum.
- \(N_e\) = jumlah hari yang memenuhi kriteria kejadian ekstrem.
- \(R_{\text{threshold}}\) = ambang batas curah hujan yang ditentukan.
- \(I\) = fungsi indikator yang bernilai 1 jika kondisi terpenuhi dan 0 jika tidak.

Ambang batas kejadian ekstrem harus ditentukan berdasarkan tujuan analisis atau kriteria yang dapat dipertanggungjawabkan. Jika ambang batas belum ditetapkan, program tidak boleh menganggap suatu nilai sebagai hujan ekstrem tanpa penjelasan.

Data kosong, nilai negatif, tanggal tidak valid, dan tanggal duplikat perlu diperiksa sebelum analisis. Nilai yang mencurigakan tidak boleh langsung dihapus atau diperbaiki tanpa aturan yang jelas.
